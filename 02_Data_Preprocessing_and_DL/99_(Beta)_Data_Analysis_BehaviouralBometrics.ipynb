#Combined Data Analysis


```python
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import confusion_matrix, accuracy_score, precision_score, recall_score, f1_score, classification_report, roc_auc_score, roc_curve
from sklearn.preprocessing import LabelEncoder

c_data= pd.read_csv('/content/drive/MyDrive/MinorProject/Sensor Datas/Combined Datas CSV.csv')

data_csv_info = c_data.info()
data_csv_head = c_data.head()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

c_data['Timestamp'] = pd.to_datetime(c_data['Timestamp'])

c_data.hist(figsize=(12, 12))
plt.show()

plt.figure(figsize=(15, 6))
for user_id in c_data['Name'].unique():
    subset = c_data[c_data['Name'] == user_id]
    plt.plot(subset['Timestamp'], subset['Accelerometer X'], label=f'User {user_id}')
plt.legend()
plt.title('Accelerometer Data Over Time')
plt.xlabel('Timestamp')
plt.ylabel('Accelerometer X')
plt.show()
```

```python
c_data['Timestamp'] = pd.to_datetime(c_data['Timestamp'])

c_data['Relative Time'] = c_data.groupby('Name')['Timestamp'].transform(lambda x: (x - x.min()))

c_data['Relative Time'] = c_data['Relative Time'].dt.total_seconds()

plt.figure(figsize=(15, 8))
names = c_data['Name'].unique()
for name in names:
    subset = c_data[c_data['Name'] == name]
    plt.plot(subset['Relative Time'], subset['Accelerometer X'], label=f'{name}')

plt.legend()
plt.title('Accelerometer Data Over Relative Time')
plt.xlabel('Seconds from Start')
plt.ylabel('Accelerometer X')
plt.show()
```

```python
c_data['Timestamp'] = pd.to_datetime(c_data['Timestamp'])

c_data['Relative Time'] = c_data.groupby('Name')['Timestamp'].transform(lambda x: (x - x.min()))

c_data['Relative Time'] = c_data['Relative Time'].dt.total_seconds()

plt.figure(figsize=(15, 8))
names = c_data['Name'].unique()
for name in names:
    subset = c_data[c_data['Name'] == name]
    plt.scatter(subset['Relative Time'], subset['Last Touch X'], label=f'{name} Last Touch X', alpha=0.6)
    plt.scatter(subset['Relative Time'], subset['Last Touch Y'], label=f'{name} Last Touch Y', alpha=0.6)

plt.legend()
plt.title('Last Touch Data Over Relative Time')
plt.xlabel('Seconds from Start')
plt.ylabel('Last Touch Position')
plt.show()
```

```python
import pandas as pd

c_data['Timestamp'] = pd.to_datetime(c_data['Timestamp'])

window_size = 10

c_data['RollingMeanX'] = c_data.groupby('Name')['Accelerometer X'].transform(lambda x: x.rolling(window=window_size).mean().fillna(method='bfill'))
c_data['RollingMeanY'] = c_data.groupby('Name')['Accelerometer Y'].transform(lambda x: x.rolling(window=window_size).mean().fillna(method='bfill'))
c_data['RollingMeanZ'] = c_data.groupby('Name')['Accelerometer Z'].transform(lambda x: x.rolling(window=window_size).mean().fillna(method='bfill'))

c_data['StdDevX'] = c_data.groupby('Name')['Accelerometer X'].transform(lambda x: x.rolling(window=window_size).std().fillna(method='bfill'))
c_data['StdDevY'] = c_data.groupby('Name')['Accelerometer Y'].transform(lambda x: x.rolling(window=window_size).std().fillna(method='bfill'))
c_data['StdDevZ'] = c_data.groupby('Name')['Accelerometer Z'].transform(lambda x: x.rolling(window=window_size).std().fillna(method='bfill'))
print(c_data[['RollingMeanX', 'RollingMeanY', 'RollingMeanZ']].head())
print(c_data[['StdDevX', 'StdDevY', 'StdDevZ']].head())

```

```python
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score
X = c_data[['RollingMeanX', 'RollingMeanY', 'RollingMeanZ', 'StdDevX', 'StdDevY', 'StdDevZ']]  # add more features if needed
y = c_data['Name']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

model = RandomForestClassifier(n_estimators=100, random_state=42)

model.fit(X_train, y_train)
y_pred = model.predict(X_test)
accuracy = accuracy_score(y_test, y_pred)
print(f'Accuracy: {accuracy:.2f}')
```

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix

# Evaluate the model
accuracy = accuracy_score(y_test, y_pred)
precision = precision_score(y_test, y_pred, average='macro')
recall = recall_score(y_test, y_pred, average='macro')
f1 = f1_score(y_test, y_pred, average='macro')
conf_matrix = confusion_matrix(y_test, y_pred)

print(f'Accuracy: {accuracy:.2f}')
print(f'Precision: {precision:.2f}')
print(f'Recall: {recall:.2f}')
print(f'F1-Score: {f1:.2f}')
print("Confusion Matrix:\n", conf_matrix)


```

```python
# Compute confusion matrix
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.metrics import confusion_matrix

conf_matrix = confusion_matrix(y_test, y_pred)
labels = model.classes_

plt.figure(figsize=(10, 7))
sns.heatmap(conf_matrix, annot=True, fmt='d', cmap='Blues', xticklabels=labels, yticklabels=labels)
plt.title('Confusion Matrix')
plt.xlabel('Predicted Labels')
plt.ylabel('True Labels')
plt.show()
```

```python
from sklearn.metrics import roc_curve, auc
from sklearn.preprocessing import label_binarize
import numpy as np

y_test_binarized = label_binarize(y_test, classes=np.unique(y_test))
y_pred_proba = model.predict_proba(X_test)

n_classes = y_test_binarized.shape[1]

fpr = dict()
tpr = dict()
roc_auc = dict()

for i in range(n_classes):
    fpr[i], tpr[i], _ = roc_curve(y_test_binarized[:, i], y_pred_proba[:, i])
    roc_auc[i] = auc(fpr[i], tpr[i])

plt.figure(figsize=(10, 8))

for i in range(n_classes):
    plt.plot(fpr[i], tpr[i], label=f'ROC curve of class {labels[i]} (area = {roc_auc[i]:.2f})')

plt.plot([0, 1], [0, 1], 'k--')
plt.xlim([0.0, 1.0])
plt.ylim([0.0, 1.05])
plt.xlabel('False Positive Rate')
plt.ylabel('True Positive Rate')
plt.title('Receiver Operating Characteristic')
plt.legend(loc="lower right")
plt.show()

```

```python
from sklearn.model_selection import GridSearchCV

# Define the parameter grid
param_grid = {
    'n_estimators': [50, 100, 200],
    'max_depth': [10, 20, 30, None],
    'min_samples_split': [2, 5, 10]
}

grid_search = GridSearchCV(estimator=RandomForestClassifier(random_state=42),
                           param_grid=param_grid,
                           cv=5,
                           verbose=2,
                           n_jobs=-1,
                           scoring='accuracy')


grid_search.fit(X_train, y_train)

best_params = grid_search.best_params_
best_score = grid_search.best_score_

print("Best parameters:", best_params)
print(f"Best cross-validated accuracy: {best_score:.2f}")
```
